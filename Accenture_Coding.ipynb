```python
#Count the occurance of the character
def CountOccurance(text):
  visited=[False]*len(text)
  for i in range(len(text)):
    if(visited[i]):
      continue
    count=1
    for j in range(i+1,len(text)):
      if(text[i]==text[j]):
        count+=1
        visited[j]=True
    print(text[i],count,end='')

txt=input()
CountOccurance(txt)
```

```python
#Check the parenthesis are valid or not
def is_valid_parenthesis(user_text):
  stack=[]
  mapping={'}':'{',']':'[',')':'('}
  for ch in user_text:
    if ch in mapping.values():
      stack.append(ch)
    elif ch in mapping:
      if not stack or stack[-1]!=mapping[ch]:
        return False
      stack.pop()
  return not stack
st=input()
is_valid_parenthesis(st)
```

```python
def fruit_freq(s):
  freq={}
  for i in s:
    freq[i]=freq.get(i,0)+1
  even_count=[]
  odd_count=[]
  for count in freq.values():
    if(count%2==0):
      even_count.append(count)
    else:
      odd_count.append(count)
  max_even=max(even_count)
  max_odd=max(odd_count)
  result=abs(max_even-max_odd)
  return result
text=input()
fruit_freq(text)
```

```python
#Find the Pythagorean Triplets

def PaithogaresTriplet(num):
  for i in range(1,num+1):
    for j in range(i+1,num+1):
      c=(i**2+j**2)**0.5
      if(c<=num and c==int(c)):
        print(i,j,int(c))
num=int(input())
PaithogaresTriplet(num)
```

```python
#MAximumSubarray(Kadles theorom)
def MaximumSubArray(arr):
  current_sum=0
  max_sum=0
  for i in arr:
    current_sum=max(i,i+current_sum)
    max_sum=max(current_sum,max_sum)
  return max_sum

arr=list(map(int,input().split()))
print(MaximumSubArray(arr))
```

```python
#Conditional SubArray

def countMiniBoxes(N, X, A):
    count = 0

    for i in range(N):
        for j in range(i, N):
            if A[i] + A[j] <= X:
                count += 1

    return count


N = 4
X = 5
A = [1, 3, 5, 8]

print(countMiniBoxes(N, X, A))
```

```python
#Spiral Matrix

row,col=map(int,input().split())
matrix=[]
for i in range(row):
  matrix.append(list(map(int,input().split())))
top=0
bottom=row-1
left=0
right=col-1

while (left<=right and bottom<=top):
  for i in range(left,right+1):
    print(matrix[top][i],end=" ")
  left+=1

  for i in range(top,bottom+1):
    print(matrix[i][right],end=" ")
  right-=1

  if top<=bottom:
    for i in range(right,left-1,-1):
      print(matrix[bottom][i],end=" ")
    bottom-=1

  if(left<=right):
    for i in range(bottom, top-1,-1):
      print(matrix[i][left],end=" ")
    left+=1
```
