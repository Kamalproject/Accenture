```python
def numberofflower(input1, input2):
    transformed = []

    for i in range(input1):
        x = input2[i]


        less_count = sum(1 for val in input2 if val < x)
        greater_count = sum(1 for val in input2 if val > x)
        freq = input2.count(x)
        if less_count == greater_count and freq < 3:
            transformed.append(x)
        else:
            transformed.append(0)

    if any(val != 0 for val in transformed):
        return sum(transformed)
    else:
        return sum(input2)

input1 = int(input())
input2 = list(map(int, input().split()))

print(numberofflower(input1, input2))
```

```python
def group_by_sign(arr):
    if not arr:
        return []

    groups = [[arr[0]]]

    for num in arr[1:]:
        # Compare sign of current number with sign of last group's numbers
        same_sign = (num >= 0) == (groups[-1][-1] >= 0)
        if same_sign:
            groups[-1].append(num)
        else:
            groups.append([num])

    return groups


arr = [1, 3, -4, 7, 4]
result = group_by_sign(arr)
print(result)
```

```python
#Graph build

def Graphbuild(edges):
  graph={}
  for edge in edges:
    a,b=edge
    if a not in graph:
      graph[a]=[]
    if b not in graph:
      graph[b]=[]
    graph[a].append(b)
    graph[b].append(a)
  return graph


edges=[['i','j'],['k','l'],['m','k'],['k','l'],['o','m']]
Graphbuild(edges)
```

```python
class BankAccount():


  def __init__(self):
    print("------------------Account Creation-------------------\n")
    self.name=input("Enter Your Name: ")
    self.balance=int(input("How much amount you want to deposit(minimum 1000): "))

    print(f"Congratulations {self.name} your account has been created successfully with balance {self.balance}")

  def deposit(self):
    print("\n-------------------Deposit amount---------------------\n")
    amount=int(input("How much amount you want to depsoit: "))
    self.balance+=amount
    print(f"You deposited {amount} your balance has been updated")

  def withdraw(self):
    print("-----------------Withdraw amount---------------------\n")
    amount=int(input("Enter amount you want to withdraw: "))
    if(amount>self.balance):
      print("You don't have sufficient balance")
      return
    else:
      self.balance-=amount
      print(f"You withdraw {amount} your balance has been updated")

  def display(self):
    print(f"Your balance is {self.balance}")

bankaccount=BankAccount()

button=int(input("Press button for perform operation:\n 1 for Deposit\n 2 for Withdrawl\n 3 for Display \n" ))
match button:
  case 1:
    bankaccount.deposit()
  case 2:
    bankaccount.withdraw()
  case 3:
    bankaccount.display()
  case _:
    print("Invalid input")

```
