```python
#Find the Second largest
n=int(input("Enter the Size of the array: "))
arr=[]
for i in range(n):
  element=int(input('Enter element: '))
  arr.append(element)
max=arr[0]
min=arr[0]
;second_largest=arr[0]
for j in range(n):
  if(arr[j]>min and arr[j]<max):
    second_largest=arr[j]
print(second_largest)
```
